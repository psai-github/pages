### Warm-up 1: Comparisons Create Booleans

SFI backend records contain values such as product names, categories, spec numbers, and record counts. Comparison operators turn checks on those values into `True` or `False`.

| Python | Meaning | SFI example |
| --- | --- | --- |
| `==` | equal to | `spec_number == "1.1"` |
| `!=` | not equal to | `category != "Street Car"` |
| `>` | greater than | `part_count > 0` |
| `<` | less than | `part_count < 100` |
| `>=` | greater than or equal to | `part_count >= 1` |
| `<=` | less than or equal to | `part_count <= 100` |

```python
# CODE_RUNNER: Warm-up 1 - Compare SFI Part Data

spec_number = "1.1"
category = "Auto Racing"
part_count = 12

print("Spec is 1.1:", spec_number == "1.1")
print("Category is not Street Car:", category != "Street Car")
print("At least one part record exists:", part_count >= 1)
```

Remember:
- `spec_number = "1.1"` assigns a value.
- `spec_number == "1.1"` compares two values.

### Warm-up 2: `and`, `or`, and `not`

Backend validation often needs more than one check at the same time.

| Idea | Python | SFI backend meaning |
| --- | --- | --- |
| both must be true | `and` | product name **and** spec number are present |
| at least one is true | `or` | category is Auto Racing **or** Drag Racing |
| reverse true/false | `not` | record is **not** a duplicate |

```python
# CODE_RUNNER: Warm-up 2 - Combine SFI Backend Checks

has_product_name = True
has_spec_number = True
duplicate_spec = False

print("Required fields present:", has_product_name and has_spec_number)
print("At least one field present:", has_product_name or has_spec_number)
print("Spec is new:", not duplicate_spec)
```

### Warm-up 3: Build a Compound SFI Rule

A backend rule becomes more useful when several Boolean checks are combined into one expression.

```python
# CODE_RUNNER: Warm-up 3 - SFI Record Validation

has_product_name = True
has_spec_number = True
valid_category = True
duplicate_spec = False

basic_record_is_valid = (
    has_product_name
    and has_spec_number
    and valid_category
    and not duplicate_spec
)

print("Basic SFI record is valid:", basic_record_is_valid)
```

### Warm-up 4: Booleans Control Backend Decisions

An `if` statement can use a Boolean result to decide whether the backend should return or reject an SFI part record.

```python
# CODE_RUNNER: Warm-up 4 - SFI Backend Decision

part_in_database = True
record_archived = False

can_return_record = part_in_database and not record_archived

if can_return_record:
    print("Return the SFI part record")
else:
    print("Do not return the SFI part record")
```
