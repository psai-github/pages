A **random value** is selected from a set or range of possible values. Python is the main language for this lesson. JavaScript and College Board pseudocode are only used to compare the same idea in different forms.

## 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to use Python's `random` module to generate random integers, choose random list items, and use random results in a program.

**Success Criteria:** You can identify every possible output of a random expression, choose the correct range, and explain why repeated results are allowed.

### Tech Talk 1: Random Integers

Python's built-in `random` module gives us tools for random values.

```python
# CODE_RUNNER: Run the random integer example

import random

die_roll = random.randint(1, 6)
print("Die roll:", die_roll)
```

`random.randint(a, b)` can return any whole number from `a` through `b`, including both endpoints.

For a normal die, the possible results are `1`, `2`, `3`, `4`, `5`, and `6`.

### Tech Talk 2: Random Choices

Python can choose directly from a list.

```python
# CODE_RUNNER: Run the random choice example

import random

activities = ["study", "stretch", "walk", "get water"]
pick = random.choice(activities)

print("Random activity:", pick)
```

The result must be one item from the list.

The same random-integer idea looks like this in the three forms:

| Form | Random integer from 1 through 6 |
| --- | --- |
| Python | `random.randint(1, 6)` |
| JavaScript | `Math.floor(Math.random() * 6) + 1` |
| College Board pseudocode | `RANDOM(1, 6)` |

### Tech Talk 3: Use Random Values in Decisions

A program can generate a random value and then use that result in an `if` statement.

```python
# CODE_RUNNER: Run the random decision example

import random

flip = random.randint(1, 2)

if flip == 1:
    print("Heads")
else:
    print("Tails")
```

Random does **not** mean every possible result appears once before a repeat. Repeats are normal.

### Popcorn Hack 4 / Homework: SFI Backend QA Simulator

Build a reusable randomized QA simulator for the SFI automotive backend.

The simulator works with structured SFI part records, backend actions, existing spec numbers, and repeated test runs. Its behavior should reflect realistic backend outcomes for searching, creating, updating, and removing part records, including duplicate-spec handling during creation.

The final runner should produce clear test output showing which record and backend action were selected and what result occurred.

```python
# CODE_RUNNER: Popcorn Hack 4 / Homework - SFI Backend QA Simulator

import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update"
]

existing_spec_numbers = ["1.1", "2.1"]

for test_number in range(1, 6):
    selected_part = parts[0]
    selected_action = actions[0]

    spec_number = selected_part["spec_number"]

    if selected_action == "GET/search":
        result = "RETURN matching record"
    elif selected_action == "POST/create":
        duplicate_spec = spec_number in existing_spec_numbers
        result = "REJECT duplicate" if duplicate_spec else "ACCEPT new record"
    elif selected_action == "PUT/update":
        result = "VALIDATE then update record"
    else:
        result = "DELETE behavior not implemented"

    print(
        "Test", test_number,
        "|", selected_action,
        "| Spec", spec_number,
        "|", selected_part["product_name"],
        "|", result
    )
```
