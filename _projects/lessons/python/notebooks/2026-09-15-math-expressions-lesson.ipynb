## **Math Expressions Lesson**

### **Part 1 — The Basics: What is a Math Expression?**

**Why this matters:** Every program that calculates a price, a score, a distance, or a time is built on math expressions. A math expression combines numbers and operators so the computer evaluates them into a single new value — without expressions, a program can only store numbers, it can never actually *do* anything with them.

Python uses the same basic operators as a calculator:

<table>
  <thead>
    <tr><th>Operator</th><th>Meaning</th></tr>
  </thead>
  <tbody>
    <tr><td><strong>+</strong></td><td>Addition</td></tr>
    <tr><td><strong>-</strong></td><td>Subtraction</td></tr>
    <tr><td><strong>*</strong></td><td>Multiplication</td></tr>
    <tr><td><strong>/</strong></td><td>Division</td></tr>
  </tbody>
</table>

**Example:** `5 * 4` evaluates to `20`.

```python
# 1. Basic Arithmetic
total_cost = 15 + 20
print("Addition Result:", total_cost)

discount = 50 - 15
print("Subtraction Result:", discount)

# 2. Multiplication and Division
area = 10 * 5
print("Multiplication Result:", area)

half_pizza = 8 / 2
print("Division Result:", half_pizza)

# 3. Multiple Operations (order of operations applies: PEMDAS)
complex_math = (10 + 5) * 2
print("Complex Math Result:", complex_math)
```

### **Part 2 — Going Deeper: Reverse Polish Notation (RPN)**

Most expressions you write use **infix notation** — the operator sits between the numbers, like `5 + 3`. Computers don't have to evaluate math that way.

In the 1920s, Polish logician **Jan Łukasiewicz** invented **Polish notation**, writing the operator *before* the numbers (`+ 5 3`). In the 1950s, computer scientist **Charles Hamblin** flipped the order to create **Reverse Polish Notation (RPN)**, where the operator comes *after* the numbers (`5 3 +`).

| Normal math | RPN |
|---|---|
| `2 + 3` | `2 3 +` |
| `(5 + 3) * 2` | `5 3 + 2 *` |

RPN removes the need for parentheses and order-of-operations rules — every expression can be evaluated left to right using a **stack**: push numbers on, and when you hit an operator, pop the last two numbers off, apply the operator, and push the result back on.

RPN is still used today in some scientific and financial calculators because it's fast and never ambiguous.
