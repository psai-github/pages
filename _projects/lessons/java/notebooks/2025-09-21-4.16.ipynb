# AP CSA — Topic 4.16: Recursion

**Learning Objective (4.16.A):** Determine the result of calling recursive methods.

## Key Concepts

**What is a recursive method?**
A method that calls itself in order to solve a problem. (EK 4.16.A.1)

**Two required parts of every recursive method:**
1. **Base case** — a condition that stops the recursion. Without this, the method would call itself forever and cause a `StackOverflowError`.
2. **Recursive call** — the method calling itself, usually with a different (smaller/simpler) input that moves toward the base case.

## Recursion vs. Loops

- Recursion is just another form of repetition — anything a loop can do, recursion can also do, and vice versa. (EK 4.16.A.3)
- Any recursive solution can be rewritten as an iterative (loop-based) solution, and any loop can be rewritten recursively.


## Local Variables in Recursion

- Each recursive call gets its **own separate copy** of local variables and parameters. (EK 4.16.A.2)
- These parameter values track the progress of the recursion, similar to how a loop control variable (like `i` in a `for` loop) tracks progress in a loop.
- This is why tracing recursive calls often involves drawing a "stack" of calls — each call has its own state.

## ⚠️ Exam Scope Note

You will **not** be asked to *write* recursive code on the AP exam. You only need to:
- Trace through given recursive code
- Determine what value it returns / what it outputs
- Understand base cases and recursive calls conceptually

## Example to Practice Tracing

Trace `mystery(3)`:
- `mystery(3)` → `3 + mystery(2)`
- `mystery(2)` → `2 + mystery(1)`
- `mystery(1)` → `1 + mystery(0)`
- `mystery(0)` → returns `0` (base case hit)
- Unwind: `1+0=1`, `2+1=3`, `3+3=6` → **returns 6**

```python
def mystery(n):
    if n <= 0:          # base case
        return 0
    else:
        return n + mystery(n - 1)  # recursive call

print(mystery(3))  # should output 6
```
