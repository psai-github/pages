## 2. Lesson Plan

**Learning Objective:** Identify the base case and recursive call, trace recursive execution, and write recursive methods that move toward a base case.

**Success Criteria:** You can explain when recursion stops, describe how calls are stored on the call stack, predict a recursive method's output, and avoid a call that repeats the same argument.

### Tech Talk (5 minutes)

A **recursive method** calls itself to solve a smaller version of the same problem. Every correct recursive method needs two parts:

1. A **base case** that stops the recursion.
2. A **recursive case** that calls the method again with a simpler argument.

```java
public static int factorial(int n) {
    if (n <= 1) {              // base case
        return 1;
    }
    return n * factorial(n - 1); // recursive case
}
```

For `factorial(3)`, the calls are paused like this:

```text
factorial(3)
  3 * factorial(2)
      2 * factorial(1)
          1  <- base case
```

The calls then return in reverse order: `1`, `2`, and finally `6`.

The recursive argument must move toward the base case. A call such as `factorial(n)` inside `factorial(n)` never makes progress and eventually causes a `StackOverflowError`.

## 3. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
|---|---|---|
| **Recursion** | A method solving a problem by calling itself. | `factorial(n - 1)` |
| **Base case** | Condition that stops further calls. | `if (n <= 1)` |
| **Recursive case** | Part that calls the method again. | `return n * factorial(n - 1);` |
| **Call stack** | Stack of paused method calls waiting for a result. | `factorial(3)` waits for `factorial(2)` |
| **Unwinding** | Returning results from the base case back upward. | `1 → 2 → 6` |
| **StackOverflowError** | Runtime error caused by recursion that does not stop. | A call that keeps the same argument |

### Recursion Checklist

Before running a recursive method, ask:

- What is the base case?
- What does the base case return or print?
- What is the recursive call?
- Does the argument get closer to the base case?
- What values return while the call stack unwinds?

### Recursion Compared with Loops

- A loop repeats through a condition and update.
- Recursion repeats through method calls and smaller arguments.
- Both need progress and a stopping condition.

## 4. Code Examples

### A. Simple: Countdown

This method prints before making its recursive call, so the output counts down toward the base case.

````java
// CODE_RUNNER: Change the starting number and predict the output
public class RecursionCountdown {
    public static void countdown(int number) {
        if (number == 0) {
            System.out.println("Liftoff!");
        } else {
            System.out.println(number);
            countdown(number - 1);
        }
    }

    public static void main(String[] args) {
        countdown(3);
    }
}

RecursionCountdown.main(null);
````

### B. Intermediate: Return Values and Unwinding

The method reaches the base case first, then multiplies values as the paused calls return.

````java
// CODE_RUNNER: Predict factorial(4), then run it
public class RecursiveFactorial {
    public static int factorial(int number) {
        if (number <= 1) {
            return 1;
        }
        return number * factorial(number - 1);
    }

    public static void main(String[] args) {
        System.out.println("factorial(4) = " + factorial(4));
    }
}

RecursiveFactorial.main(null);
````

### C. Complex: Recursive Digit Sum

Each recursive call removes the last digit with integer division. The base case is reached when no digits remain.

````java
// CODE_RUNNER: Change the number and trace the recursive calls
public class RecursiveDigitSum {
    public static int sumDigits(int number) {
        if (number < 10) {
            return number;
        }

        int lastDigit = number % 10;
        int remainingDigits = number / 10;
        return lastDigit + sumDigits(remainingDigits);
    }

    public static void main(String[] args) {
        int number = 1234;
        System.out.println("Digit sum of " + number + ": " + sumDigits(number));
    }
}

RecursiveDigitSum.main(null);
````

### Homework Hack

Build a recursive power method. The base case is `exponent == 0`, because every nonzero number to the zero power is `1`. Each recursive call should reduce the exponent by one. Test your method with at least two inputs and explain how the argument moves toward the base case.

````java
// CODE_RUNNER: Complete the TODO, then test different values
public class RecursivePower {
    public static int power(int base, int exponent) {
        if (exponent == 0) {
            return 1;
        }

        // The exponent decreases, so the call moves toward exponent == 0.
        return base * power(base, exponent - 1);
    }

    public static void main(String[] args) {
        System.out.println("2^5 = " + power(2, 5));
        System.out.println("3^3 = " + power(3, 3));
    }
}

RecursivePower.main(null);
````
