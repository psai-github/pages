## 2. Lesson Plan

**Learning Objective:** Identify the base case and recursive call, trace recursive execution, and write recursive methods that move toward a base case.

**Success Criteria:** You can explain when recursion stops, describe call-stack frames, predict output, and repair a recursive method whose argument does not progress.

### Tech Talk (5 minutes)

Our running scenario is a launch mission. The countdown calls itself with a smaller number until it reaches zero:

```java
public static void countdown(int number) {
    if (number == 0) {
        System.out.println("Liftoff!");
    } else {
        System.out.println(number);
        countdown(number - 1);
    }
}
```

For `countdown(3)`, frames stack up as `3`, `2`, `1`, and `0`. The base case at `0` stops new calls; then the paused calls unwind.

A recursive method needs:

1. A **base case** that stops recursion.
2. A **recursive case** that calls the method with a simpler argument.
3. **Progress** toward the base case.

A call such as `countdown(number)` makes no progress and eventually causes a `StackOverflowError`.

## 3. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
|---|---|---|
| **Recursion** | A method solving a problem by calling itself. | `power(base, exponent - 1)` |
| **Base case** | Condition that stops further calls. | `if (exponent == 0)` |
| **Recursive case** | Part that calls the method again. | `return base * power(base, exponent - 1);` |
| **Call stack** | Stack of paused method calls waiting for a result. | `power(2, 3)` waits for `power(2, 2)` |
| **Unwinding** | Returning results from the base case back upward. | `1 → 2 → 4 → 8` |
| **StackOverflowError** | Runtime error caused by recursion that does not stop. | An unchanged argument |

### Recursion Checklist

- What is the base case?
- What does it return or print?
- What is the recursive call?
- Does its argument move toward the base case?
- What happens while the calls unwind?

## 4. Code Examples

### A. Mission Countdown

This runner prints before the recursive call, so the mission counts down toward liftoff.

````java
// CODE_RUNNER: Change the starting number and predict the mission output
public class MissionCountdown {
    public static void countdown(int number) {
        if (number == 0) {
            System.out.println("Liftoff!");
        } else {
            System.out.println("T-minus " + number);
            countdown(number - 1);
        }
    }

    public static void main(String[] args) {
        countdown(3);
    }
}

MissionCountdown.main(null);
````

### B. Call Stack and Fuel Calculation

The calls reach the base case first. Then the paused calls return in reverse order, calculating the mission's fuel value.

````java
// CODE_RUNNER: Predict factorial(4), then run the fuel calculation
public class MissionFuel {
    public static int factorial(int number) {
        if (number <= 1) {
            return 1;
        }
        return number * factorial(number - 1);
    }

    public static void main(String[] args) {
        System.out.println("Fuel units for mission 4: " + factorial(4));
    }
}

MissionFuel.main(null);
````

### C. Launch-Code Checksum

Each call removes the last digit. The base case is reached when one digit remains, then the checksum unwinds back through the call stack.

````java
// CODE_RUNNER: Change the launch code and predict its checksum
public class LaunchCodeChecksum {
    public static int sumDigits(int number) {
        if (number < 10) {
            return number;
        }
        return (number % 10) + sumDigits(number / 10);
    }

    public static void main(String[] args) {
        int launchCode = 2026;
        System.out.println("Checksum: " + sumDigits(launchCode));
    }
}

LaunchCodeChecksum.main(null);
````
