# 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to identify the base case and recursive call in a recursive method, trace its execution call-by-call, and determine its final returned or printed value — without needing to write recursive code yourself.

**Success Criteria:** Given an unfamiliar recursive method, you can label its base case and recursive call, trace at least 3 levels of calls on paper or in comments, and correctly predict the output.

## Tech Talk & Introduction (3 minutes)

A recursive method is a method that calls itself. Recursion is another form of repetition — anything a loop can do, recursion can also do.

**The Core Rule:** Every recursive method needs a **base case** (stops the recursion) and a **recursive call** (moves closer to that base case). Miss either one and the method either never runs the intended logic, or runs forever until it crashes.

- ✅ Do this: check for a base case *before* trusting the recursive call
- ❌ Don't do this: assume a method is "recursive" just because it calls itself somewhere inside its own body

> A method with no base case is **not** valid recursion, even if it technically calls itself — it will never stop, and will throw a `StackOverflowError`.

### Reference Guide

| Term | Meaning | What to look for when tracing |
|---|---|---|
| Base case | Condition that stops recursion | Usually an `if` that `return`s a fixed value with no further recursive call |
| Recursive call | The method calling itself | Look at what argument it passes — it should move *toward* the base case |
| Call stack | Stack of paused calls waiting on the one below them | Each call has its own copy of parameters/local variables |
| Unwinding | Returning values back up the stack | Happens after the base case is hit, in reverse call order |
| `StackOverflowError` | Runtime error from missing/unreachable base case | Signals infinite recursion |

### Code Examples

**A. Simple — identify the parts**

```java
public class Mystery {
    public static int mystery(int n) {
        if (n <= 0) {                    // base case
            return 0;
        } else {
            return n + mystery(n - 1);   // recursive call
        }
    }

    public static void main(String[] args) {
        System.out.println(mystery(3));
    }
}

Mystery.main(null);
```

**B. Intermediate — trace it**

```java
public class CountDown {
    public static int countDown(int n) {
        if (n == 0) {
            System.out.println("Liftoff!");
            return 0;
        } else {
            System.out.println(n);
            return countDown(n - 1);
        }
    }

    public static void main(String[] args) {
        countDown(3);
        // prints: 3, 2, 1, Liftoff!
    }
}

CountDown.main(null);
```

**C. Complex — spot the bug**

```java
public class Broken {
    public static int broken(int n) {
        if (n <= 0) {
            return 0;
        } else {
            return n + broken(n);   // ⚠️ passes n, not n - 1
        }
    }

    public static void main(String[] args) {
        // broken(3) never reaches the base case -> StackOverflowError
        System.out.println(broken(3));
    }
}

// Uncomment to see the crash for yourself:
// Broken.main(null);
```

### Class Discussion — Take Notes (counts toward hack points!)

1. Trace `mystery(3)` call by call. What value does each call return before it "unwinds" back up?
2. What would happen if the base case were removed from `mystery`? Why?
3. `mystery(n)` could also be written using a `for` loop instead of recursion. Sketch what that loop would look like.
4. **Most common mistake:** What is wrong with a recursive call that passes the same value of `n` every time instead of `n - 1`? Why does this cause a problem?

> ⚠️ **Exam Scope:** Writing recursive code is outside the scope of the AP CSA exam. You will only be asked to trace given recursive code, determine the value returned/printed, and identify the base case and recursive call in a method.

### Local Variables and the Call Stack

- Each recursive call gets its own copy of local variables and parameters — they don't share state with other calls.
- Parameter values track progress through the recursion, the same way a loop control variable (like `i`) tracks progress through a loop.
- This is why tracing recursion is often drawn as a stack of calls, each waiting on the one below it to finish.

**Homework Hack**

Task: The method below is supposed to sum the digits of a number but has a bug in its recursive call. Identify the base case, identify what's wrong with the recursive call, fix it, and trace `sumDigits(123)` step-by-step showing each call's return value.

```java
// UI_RUNNER: Recursion Homework Base
public class SumDigits {
    public static int sumDigits(int n) {
        if (n == 0) {
            return 0;
        } else {
            return (n % 10) + sumDigits(n);   // bug: never shrinks n
        }
    }

    public static void main(String[] args) {
        // This currently causes a StackOverflowError -- find and fix the bug!
        System.out.println(sumDigits(123));
    }
}

// Uncomment once fixed:
// SumDigits.main(null);
```
