## 2. Lesson Plan

**Learning Objective:** Use standard `for`, `while`, and enhanced `for` loops to access every element in a one-dimensional array.

**Success Criteria:** You can traverse an entire array, explain why valid indices range from `0` to `length - 1`, repair a `<=` bounds error, and choose an appropriate loop.

### Tech Talk (5 minutes)

Our running scenario is a game leaderboard:

```java
int[] scores = {982, 915, 877, 844};
```

The array has length `4`. Its valid indices are `0`, `1`, `2`, and `3`; index `4` is outside the array.

A complete index-based traversal uses `<`:

```java
for (int rank = 0; rank < scores.length; rank++) {
    System.out.println(scores[rank]);
}
```

A `while` loop can traverse the same leaderboard when you control the index yourself. An enhanced `for` loop is useful when you only need each score and not its index.

The loop condition must be `rank < scores.length`, not `rank <= scores.length`. The latter asks for one nonexistent rank.

## 3. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
|---|---|---|
| **Array** | Fixed-size collection of values with the same type. | `int[] scores = {980, 900};` |
| **Index** | Position of an element in an array. | `scores[0]` |
| **Length** | Number of elements in an array. | `scores.length` |
| **Traversal** | Visiting every element in an array. | A loop from `0` to `length - 1` |
| **Bounds error** | Attempting to access an invalid index. | `scores[scores.length]` |
| **Enhanced for loop** | Loop that visits each element directly. | `for (int score : scores)` |

### Choosing a Loop

- Need the rank or index? Use a standard `for` loop.
- Need to update elements? Use a standard `for` loop.
- Need manual condition control? Use a `while` loop.
- Only need each score? Use an enhanced `for` loop.

## 4. Code Examples

### A. Print the Leaderboard with a Standard `for` Loop

This runner declares its own scores, prints every valid index, and gives you a complete traversal to modify.

````java
// CODE_RUNNER: Change a score, run it, and predict the updated leaderboard
public class LeaderboardTraversal {
    public static void main(String[] args) {
        int[] scores = {982, 915, 877, 844};

        for (int rank = 0; rank < scores.length; rank++) {
            System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
        }
    }
}

LeaderboardTraversal.main(null);
````

### B. Compare `while` and Enhanced `for`

The `while` loop tracks the rank. The enhanced `for` loop displays values when the rank is not needed.

````java
// CODE_RUNNER: Compare the two ways to display the leaderboard
public class LeaderboardLoopChoices {
    public static void main(String[] args) {
        int[] scores = {982, 915, 877, 844};

        System.out.println("While loop with ranks:");
        int rank = 0;
        while (rank < scores.length) {
            System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
            rank++;
        }

        System.out.println("Enhanced for loop with values only:");
        for (int score : scores) {
            System.out.println(score);
        }
    }
}

LeaderboardLoopChoices.main(null);
````

### C. Catch the `<=` Bounds Error

This runner intentionally uses `<=` so you can observe the error safely. After running it, change the condition to `<` and run it again.

````java
// CODE_RUNNER: Run once to observe the bounds error, then change <= to <
public class LeaderboardBoundsCheck {
    public static void main(String[] args) {
        int[] scores = {982, 915, 877, 844};

        try {
            for (int rank = 0; rank <= scores.length; rank++) {
                System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
            }
        } catch (ArrayIndexOutOfBoundsException error) {
            System.out.println("Bounds error: index " + scores.length + " does not exist.");
        }
    }
}

LeaderboardBoundsCheck.main(null);
````

### Popcorn Hack (In-Class)

Create a six-player leaderboard. Print every player's rank and score with a standard `for` loop. Then use an enhanced `for` loop to print the score values without ranks. Replace the sample scores with your own.

````java
// CODE_RUNNER: Replace the sample scores and run your leaderboard
public class LeaderboardPopcorn {
    public static void main(String[] args) {
        int[] scores = {760, 845, 912, 688, 934, 801};

        System.out.println("Ranked players:");
        for (int rank = 0; rank < scores.length; rank++) {
            System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
        }

        System.out.println("Score values:");
        for (int score : scores) {
            System.out.println(score);
        }
    }
}

LeaderboardPopcorn.main(null);
````

### Homework Hack

The starter runner has one intentional bug: its loop uses `<=`. Run it to see the caught bounds message, then change only the condition to `<`. In your markdown response, explain why an array of length `5` has valid indices `0` through `4`.

````java
// CODE_RUNNER: Repair <= to <, then run the corrected traversal
public class LeaderboardHomework {
    public static void main(String[] args) {
        int[] scores = {710, 865, 903, 947, 981};

        try {
            for (int rank = 0; rank <= scores.length; rank++) {
                System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
            }
        } catch (ArrayIndexOutOfBoundsException error) {
            System.out.println("Repair needed: the loop tried index " + scores.length + ".");
        }
    }
}

LeaderboardHomework.main(null);
````
