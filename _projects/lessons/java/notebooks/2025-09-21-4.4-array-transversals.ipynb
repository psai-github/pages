### Tech Talk (5 minutes)

An array stores multiple values of the same type. Each value has an index.

```java
int[] scores = {82, 91, 77, 95};
```

This array has a length of `4`, but its valid indices are `0`, `1`, `2`, and `3`.

The standard traversal pattern uses `<`:

```java
for (int index = 0; index < scores.length; index++) {
    System.out.println(scores[index]);
}
```

The condition must be `index < scores.length`, not `index <= scores.length`. The index `scores.length` is one position past the end of the array.

A `while` loop can perform the same traversal:

```java
int index = 0;
while (index < scores.length) {
    System.out.println(scores[index]);
    index++;
}
```

An enhanced `for` loop visits each value directly:

```java
for (int score : scores) {
    System.out.println(score);
}
```

Use an index-based loop when you need the index or need to change array elements. Use an enhanced `for` loop when you only need each value.

## 3. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
|---|---|---|
| **Array** | Fixed-size collection of values with the same type. | `int[] scores = {80, 90};` |
| **Index** | Position of an element in an array. | `scores[0]` |
| **Length** | Number of elements in an array. | `scores.length` |
| **Traversal** | Visiting every element in an array. | A loop from `0` to `length - 1` |
| **Bounds error** | Attempting to access an invalid index. | `scores[scores.length]` |
| **Enhanced for loop** | Loop that visits each element directly. | `for (int score : scores)` |

### Valid Array Indices

For an array with length `n`:

- First valid index: `0`
- Last valid index: `n - 1`
- Invalid index after the end: `n`

### Choosing a Loop

- Need the index? Use a standard `for` loop.
- Need to update elements? Use a standard `for` loop.
- Need a condition-controlled loop? Use a `while` loop.
- Only need each value? Use an enhanced `for` loop.

## 4. Code Examples

### A. Standard `for` Loop

The loop visits every valid index exactly once.

````python
// CODE_RUNNER: Run it, then change an element and predict the output
public class StandardArrayTraversal {
    public static void main(String[] args) {
        int[] scores = {82, 91, 77, 95};

        for (int index = 0; index < scores.length; index++) {
            System.out.println("Index " + index + ": " + scores[index]);
        }
    }
}

StandardArrayTraversal.main(null);
````

### B. `while` and Enhanced `for` Loops

Both examples below visit every element, but they provide different access.

````python
// CODE_RUNNER: Compare the two traversal styles
public class OtherArrayTraversals {
    public static void main(String[] args) {
        int[] numbers = {10, 20, 30, 40};

        System.out.println("While loop:");
        int index = 0;
        while (index < numbers.length) {
            System.out.println(numbers[index]);
            index++;
        }

        System.out.println("Enhanced for loop:");
        for (int number : numbers) {
            System.out.println(number);
        }
    }
}

OtherArrayTraversals.main(null);
````

### C. Even and Odd Indices

These loops intentionally skip elements, so they are **not complete traversals**. They are useful when a task asks for selected indices.

````python
// CODE_RUNNER: Observe which indices are selected
public class SelectedArrayIndices {
    public static void main(String[] args) {
        int[] values = {10, 20, 30, 40, 50, 60};

        System.out.println("Even indices:");
        for (int index = 0; index < values.length; index += 2) {
            System.out.println("Index " + index + ": " + values[index]);
        }

        System.out.println("Odd indices:");
        for (int index = 1; index < values.length; index += 2) {
            System.out.println("Index " + index + ": " + values[index]);
        }
    }
}

SelectedArrayIndices.main(null);
````
