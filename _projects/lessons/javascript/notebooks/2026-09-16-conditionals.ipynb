## Introduction (1-3 minutes)

Conditionals allow dynamic responses depending on whether a condition is true or false. Types include `if`, `if...else`, `else if`, and `switch`.

Imagine your morning alarm rings. **IF** it is a school weekday, check whether you have a test. If you do, get up and study; otherwise, snooze. **ELSE**, if it is a weekend, check whether you have homework. If you do, get up and do it; otherwise, go back to sleep.

See how you are making decisions **inside** other decisions? That is a **nested conditional**!
