```python

.switch {
    position: relative;
    display: flex;
    width: 54px;
    height: 31px;
    padding: 2px 2px 2px 6px;
    justify-content: flex-end;
    align-items: center;

    input[type="checkbox"] {
        opacity: 0;
        width: 0;
        height: 0;

        &:checked+.slider {
            background: $blue1;

            &.green {
                background: $green1;
            }
        }

        &:focus+.slider {
            box-shadow: 0 0 1px $blueShadow1;
        }

        &:checked+.slider:before {
            transform: translateX(27px);
        }
    }
}

.slider {
    position: absolute;
    cursor: pointer;
    inset: 0;
    border-radius: 100px;
    background: $gray4;
    background-blend-mode: overlay, luminosity;
    transition: 0.4s;

    &:before {
        content: "";
        position: absolute;
        width: 27px;
        height: 27px;
        left: 4px;
        bottom: 4px;
        border-radius: 50%;
        background: #fff;
        box-shadow: 0px 0px 0px 1px rgba(0, 0, 0, 0.04),
            0px 3px 8px 0px rgba(0, 0, 0, 0.15), 0px 3px 1px 0px rgba(0, 0, 0, 0.06);
        transition: 0.4s;
    }
}
```
