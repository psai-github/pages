### What Are Toggles?

You probably use toggles every day without even realizing it! Toggles are used in apps, websites, and devices to quickly turn settings on or off.

<img src="images/capstone/toggles.png" width="200">

This image shows a settings screen with two toggles. Battery percentage is turned on, shown by the green switch, while low power mode is turned off, shown by the gray switch. This is a common example of how toggles are used to control settings.
