## Switches (refactor Toggles for open-coding)


### What Are Toggles?

You probably use toggles every day without even realizing it! Toggles are used in apps, websites, and devices to quickly turn settings on or off.

<img src="{{site.baseurl}}/assets/images/capstone/toggles.png" width="200">

This image shows a settings screen with two toggles. Battery percentage is turned on, shown by the green switch, while low power mode is turned off, shown by the gray switch. This is a common example of how toggles are used to control settings.



### What are they good for?

### What Are Toggles Good For?

Toggles let users quickly turn a setting on or off. They are commonly used in apps and websites for things like:

- Notifications
- Dark Mode
- Privacy settings
- Battery settings
- Website features

Toggles work best when there are two choices, such as on/off or enabled/disabled.



### Code Breakdown 

.switch {
    position: relative;
    display: flex;
    width: 54px;
    height: 31px;
    padding: 2px 2px 2px 6px;
    justify-content: flex-end;
    align-items: center;

    input[type="checkbox"] {
        opacity: 0;
        width: 0;
        height: 0;

        &:checked+.slider {
            background: $blue1;

            &.green {
                background: $green1;
            }
        }

        &:focus+.slider {
            box-shadow: 0 0 1px $blueShadow1;
        }

        &:checked+.slider:before {
            transform: translateX(27px);
        }
    }
}

.slider {
    position: absolute;
    cursor: pointer;
    inset: 0;
    border-radius: 100px;
    background: $gray4;
    background-blend-mode: overlay, luminosity;
    transition: 0.4s;

    &:before {
        content: "";
        position: absolute;
        width: 27px;
        height: 27px;
        left: 4px;
        bottom: 4px;
        border-radius: 50%;
        background: #fff;
        box-shadow: 0px 0px 0px 1px rgba(0, 0, 0, 0.04),
            0px 3px 8px 0px rgba(0, 0, 0, 0.15), 0px 3px 1px 0px rgba(0, 0, 0, 0.06);
        transition: 0.4s;
    }
}

### Code Breakdown

1. **Set Up the `.switch`
   - Set the toggle's position, size, padding, and alignment.

2. **Set Up the Checkbox
   - Select the checkbox and make it invisible while keeping it functional.

3. **Style the `.slider`
   - Create the visible toggle with its position, rounded shape, background color, and transition.

4. **Create the Slider Button
   - Use `&:before` to create the white circle and set its size, position, and shadow.

5. **Add the Checked State
   - Use `&:checked + .slider` to change the toggle's color when it is turned on.
   - The `.green` class can change the checked color to `$green1`.

6. **Move the Slider Button
   - Use `translateX(27px)` to move the white circle when the toggle is turned on.

7. **Add Focus and Transitions
   - Add a focus shadow with `&:focus + .slider`.
   - Use `transition` to make the toggle changes smooth.
