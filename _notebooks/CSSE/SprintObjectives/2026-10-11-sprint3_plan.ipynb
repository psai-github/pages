### Wayfinding Journey

Students need to keep track of serverl things in order to obtain points.  You should be asking yourself the below questions and resolving confusion on a daily basis.

1. Am I on track? (Temporal wayfinding)
  - Is my progress in line with the Calendar/Announcemnt messages?
  - Do my completion/burndown actions on OCS line up with the requirements timeline?

2. Am I working with my cohorts? (Social wayfinding)
  - Is my group active and visible within the classroom and to the teacher?
  - Do I support the needs of my group inside and outside of the classroom?

2. Can I demonstrate indicators of Success (Analytical Wayfinding)?
  - Do I mentor or advaocte for clarity on my homeworks?  Do I monitor homeworks submitted that are assigned by my group?
  - Do I initiate or respond to chat? Am I able to track my OCS analytics?
